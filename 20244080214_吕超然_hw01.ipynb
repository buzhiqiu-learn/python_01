```python
def gcd(m, n):

    # 确保算法能处理任意顺序的输入，循环中会自动交换
    while n != 0:
        m, n = n, m % n
    return m

for m, n in [(2018, 2019), (1200, 300), (5040, 60)]:
    print(f"gcd({m}, {n}) = {gcd(m, n)}")

print("\n负数情形：")
for m, n in [(-2018, 2019), (2018, -2019), (-2018, -2019)]:
    print(f"gcd({m}, {n}) = {gcd(m, n)}")
```

### 负数输入时函数的行为

函数不会报错，仍然能收敛，但**符号会跟着 `%` 走**：Python 里 `a % b` 的结果与除数 `b` 同号。当 `n` 为负时，`m % n` 得到负数，最后一步会收敛到 `-1` 而不是 `1`，所以 `gcd(2018, -2019)` 和 `gcd(-2018, -2019)` 都返回 `-1`。

这个结果**部分有意义**：返回值的绝对值仍然是正确的最大公约数（1），只是符号丢了。原因是数学上 GCD 定义为非负整数，而这段实现没有取绝对值。如果要求结果恒为非负，最后一行应改成 `return abs(m)`。

```python
import math

def factorial(k):
    #计算 k 的阶乘
    if k == 0:
        return 1
    result = 1
    for i in range(1, k + 1):
        result *= i
    return result

def euler_limit(n):
    #通过公式 (1 + 1/n)^n 近似 e。
    return (1 + 1 / n) ** n

def euler_infinite_sum(n):
    #基于公式 (1) 中的前 n 项之和（1-索引）返回 e 的近似值。
    #n = 0 时返回 0；n = 1 时返回 k=0 项（1）；n = 4 时返回 k=0,1,2,3 项之和。
    if n == 0:
        return 0.0
    total = 0.0
    for k in range(n):  # k 从 0 到 n-1
        total += 1 / factorial(k)
    return total

def euler_approx(epsilon):

    #使用公式 (1) 中的和来获得一个近似 e 的值，该值在 epsilon 内接近真实值 e。

    true_e = math.exp(1)
    current_sum = 0.0
    k = 0
    while abs(current_sum - true_e) >= epsilon:
        current_sum += 1 / factorial(k)
        k += 1
    return current_sum

def print_euler_sum_table(n):
    #打印从 k=1 到 k=n 的 euler_infinite_sum(k) 的值，每行一个。
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

# 示例测试
print("euler_limit(1000):", euler_limit(1000))          # 应接近 2.7169...
print("euler_infinite_sum(0):", euler_infinite_sum(0))  # 0.0
print("euler_infinite_sum(1):", euler_infinite_sum(1))  # 1.0
print("euler_infinite_sum(4):", euler_infinite_sum(4))  # 2.6666...
approx = euler_approx(0.001)
print("euler_approx(0.001):", approx, "误差:", abs(approx - math.exp(1)))

print("\nTable for n=5:")
print_euler_sum_table(5)
```

```python
def is_power_of_2(n):

    # 循环直到 n 为 1
    while n > 1:
        # 如果 n 不能被 2 整除，则不是 2 的幂
        if n % 2 != 0:
            return False
        n //= 2  # 整除 2
    # 循环结束，n 必定为 1，说明是 2 的幂
    return True


def is_power(b, n):
    # 推广版：n 是否为 b 的幂，不断用 b 整除 n，最后剩下 1 则是
    if b == 1:
        return n == 1          # 1 的任意次幂都是 1
    while n > 1 and n % b == 0:
        n //= b
    return n == 1


# ===== 调用演示 =====
print("--- is_power_of_2 ---")
for n in [1, 2, 3, 4, 6, 8, 16, 1024]:
    print(f"is_power_of_2({n}) = {is_power_of_2(n)}")

print("\n--- is_power(b, n) ---")
for b, n in [(2, 8), (2, 12), (3, 9), (3, 10), (5, 125), (2, 1), (1, 1), (4, 64)]:
    print(f"is_power({b}, {n}) = {is_power(b, n)}")
```
