### 字符串的乐趣

```python
#任务1
def is_palindrome(s):
    chars = [c.lower() for c in s if c != " "]
    i, j = 0, len(chars) - 1
    while i < j:
        if chars[i] != chars[j]:
            return False
        i += 1; j -= 1
    return True

for t in ["level", "kayak", "pop", "taco cat", "T A C O cat", "hello", ""]:
    print(repr(t), "->", is_palindrome(t))
```

```python
#任务2
def is_abecedarian(s):
    chars = [c.lower() for c in s if c != " "]
    for i in range(len(chars) - 1):
        if chars[i] > chars[i + 1]:
            return False
    return True

for t in ["adder", "beet", "dog", "cat", "abcd efgh xyz", "a", ""]:
    print(repr(t), "->", is_abecedarian(t))
```

```python
#任务3
def remove_vowels(s):
    vowels = set("aeiou")
    return "".join(c for c in s if c.lower() not in vowels)

for t in ["cat", "audio", "aeiou", "xyz", ""]:
    print(repr(t), "->", repr(remove_vowels(t)))
```

### 列表的乐趣

```python
#任务1
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("expected a list")
    return lst[::-1]

print(list_reverse([1, 2, 3]))
print(list_reverse([]))
print(list_reverse(["a", "b"]))
try:
    list_reverse((1, 2, 3))
except TypeError as e:
    print("TypeError:", e)
```

```python
#任务2
def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True

print(is_sorted([1, 2, 3]))
print(is_sorted([1, 3, 2]))
print(is_sorted([]))
print(is_sorted([5]))
print(is_sorted((1, 2, 3)))
print(is_sorted("abc"))
```

### 作为向量的元组

```python
#任务1
import numbers

def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("t must be a tuple")
    if not isinstance(s, numbers.Number):
        raise TypeError("s must be a number")
    if not all(isinstance(x, numbers.Number) for x in t):
        raise TypeError("elements of t must be numbers")
    return tuple(x * s for x in t)

print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.5, 2.5), 2))
try: vec_scalar_mult([1, 2], 2)
except TypeError as e: print("TypeError:", e)
try: vec_scalar_mult((1, 2), "a")
except TypeError as e: print("TypeError:", e)
```

```python
#任务2
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("both inputs must be tuples")
    if len(x) != len(y):
        raise ValueError("vectors must have same length")
    if not all(isinstance(v, numbers.Number) for v in x + y):
        raise TypeError("elements must be numbers")
    return float(sum(a * b for a, b in zip(x, y)))

print(vec_inner_product((1, 2, 3), (4, 5, 6)))
print(vec_inner_product((1, 2), (3, 4)))
try: vec_inner_product([1, 2], (3, 4))
except TypeError as e: print("TypeError:", e)
try: vec_inner_product((1, 2), (3, 4, 5))
except ValueError as e: print("ValueError:", e)
```

```python
#任务3
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    if len(mx) == 0:
        return True
    if not all(isinstance(row, tuple) for row in mx):
        return False
    n = len(mx[0])
    if any(len(row) != n for row in mx):
        return False
    for row in mx:
        for elem in row:
            if not isinstance(elem, numbers.Number):
                return False
    return True

print(check_valid_mx(((1, 2), (3, 4))))
print(check_valid_mx(((1, 2), 3)))
print(check_valid_mx(((1, 2), (3, 4, 5))))
print(check_valid_mx(((1, 2), ("a", "b"))))
print(check_valid_mx(([1, 2], [3, 4])))
print(check_valid_mx(()))
```

```python
#任务4
def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise ValueError("invalid matrix")
    if not isinstance(v, tuple):
        raise TypeError("v must be a tuple")
    if not all(isinstance(x, numbers.Number) for x in v):
        raise TypeError("elements of v must be numbers")
    n_cols = len(mx[0]) if mx else 0
    if len(v) != n_cols:
        raise ValueError("dimension mismatch")
    return tuple(sum(row[j] * v[j] for j in range(n_cols)) for row in mx)

print(mx_vec_mult(((1, 2), (3, 4)), (5, 6)))
try: mx_vec_mult(((1, 2), (3, 4)), (1, 2, 3))
except ValueError as e: print("ValueError:", e)
try: mx_vec_mult(((1, 2), 3), (1, 2))
except ValueError as e: print("ValueError:", e)
```

### 更多向量的乐趣

```python
import numbers
from functools import total_ordering

@total_ordering
class Vector:
    def __init__(self, dim, elems=None):
        if not isinstance(dim, int) or isinstance(dim, bool) or dim < 0:
            raise TypeError("dim must be a non-negative integer")
        if elems is None:
            elems = [0] * dim
        if not isinstance(elems, (list, tuple)):
            raise TypeError("elems must be a list or tuple")
        if len(elems) != dim:
            raise ValueError("dim and length of elems must match")
        if not all(isinstance(e, numbers.Number) for e in elems):
            raise TypeError("elements must be numbers")
        self.dim = dim
        self.elems = tuple(elems)

    def __repr__(self):
        return f"Vector({self.dim}, {self.elems})"

    # ---- 比较：@total_ordering 要求 __eq__ + 至少一个排序方法 ----
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("cannot compare vectors of different dimensions")
        return self.elems == other.elems

    def __lt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("cannot compare vectors of different dimensions")
        return self.elems < other.elems

    # ---- 点积 ----
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("argument must be a Vector")
        if self.dim != other.dim:
            raise ValueError("dimensions must match")
        return float(sum(a * b for a, b in zip(self.elems, other.elems)))

    # ---- 乘法：标量左/右乘、向量逐元素乘 ----
    def __mul__(self, other):
        if isinstance(other, numbers.Number):
            return Vector(self.dim, tuple(x * other for x in self.elems))
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("dimensions must match")
            return Vector(self.dim, tuple(a * b for a, b in zip(self.elems, other.elems)))
        return NotImplemented

    def __rmul__(self, other):
        return self.__mul__(other)
```

```python
v1 = Vector(2, (2, 4))
v2 = Vector(2, (2, 5))
v3 = Vector(2, (2, 4))

print(v1, v2, v3)
print("==:", v1 == v3, v1 == v2)
print("< :", v1 < v2, "> :", v2 > v1, "<=:", v1 <= v3, ">=:", v2 >= v1)
print("dot:", v1.dot(v2))

v = Vector(2, (2, 3))
w = Vector(2, (4, 5))
print("v*2 =", v * 2)
print("2*v =", 2 * v)
print("v*w =", v * w)

# 异常演示
try:
    Vector(2, (1, 2)) == Vector(3, (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)

try:
    v * Vector(3, (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)

try:
    v.dot((1, 2))
except TypeError as e:
    print("TypeError:", e)
```
