## 实验二：函数、条件、循环与递归

### 任务1

```python
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在 0 到 100 之间")
    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"

# 测试示例
try:
    print(f"95分: {score_level(95)}")
    print(f"83分: {score_level(83)}")
    print(f"70分: {score_level(70)}")
    print(f"50分: {score_level(50)}")
    score_level(101)  # 测试超出范围
except ValueError as e:
    print(f"错误: {e}")
```

```python
# 使用 for 循环
sum_for = 0
for i in range(1, 101):
    sum_for += i
print(f"for 循环计算 1-100 和: {sum_for}")

# 使用 while 循环
sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1
print(f"while 循环计算 1-100 和: {sum_while}")
```

### 任务3

```python
# 递归版
def factorial_recursive(n):
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)

# 迭代版
def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

# 比较 n = 0, 1, 5, 10 时的结果
for n in [0, 1, 5, 10]:
    print(f"n = {n}: 递归版 = {factorial_recursive(n)}, 迭代版 = {factorial_iterative(n)}")
```

### 任务4

```python
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    min_val = min(values)
    max_val = max(values)
    avg_val = sum(values) / len(values)
    return (min_val, max_val, avg_val)

# 测试
data = [3, 1, 4, 1, 5, 9, 2, 6]
min_v, max_v, avg_v = summary(data)
print(f"列表: {data}")
print(f"最小值: {min_v}, 最大值: {max_v}, 平均值: {avg_v:.2f}")
```

```python
#任务1
import os

# 查看当前工作目录
print("当前工作目录:", os.getcwd())

# 构造相对路径
data_dir = os.path.join("exp01_data")

# 创建目录（如果已存在则不报错）
os.makedirs(data_dir, exist_ok=True)
print(f"数据目录已创建/确认存在: {data_dir}")
print(f"绝对位置: {os.path.abspath(data_dir)}")
```

```python
#任务2
data_path = os.path.join("exp01_data", "students.txt")

# 写入数据
data_content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete"""

with open(data_path, "w", encoding="utf-8") as f:
    f.write(data_content)
print("文件已写入:", data_path)

# 读取文件，跳过表头
with open(data_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
    data_lines = lines[1:]  # 跳过第一行表头

print("\n读取到的数据行（跳过表头）:")
for line in data_lines:
    print(line.strip())

# 用字符串方法解析前 3 行：split(",") 拆字段
print("\n用字符串方法解析前 3 行（split）:")
for line in data_lines[:3]:
    parts = line.strip().split(",")
    print(f"  name={parts[0]!r}, score={parts[1]!r}")
```

```python
#任务3
def parse_student(line):

    parts = line.strip().split(',')

    # 检查字段数量
    if len(parts) != 2:
        raise ValueError("字段数量不正确（应为2个）")

    name, score_str = parts

    # 检查姓名是否为空
    if not name:
        raise ValueError("姓名为空")

    # 尝试将成绩转换为整数
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError("成绩不是有效整数")

    # 检查成绩范围
    if not (0 <= score <= 100):
        raise ValueError("成绩不在0~100范围内")

    return {"name": name, "score": score}

print("parse_student('Alice,85') ->", parse_student("Alice,85"))
```

```python
#任务4
valid_records = []
invalid_records = []

with open(data_path, "r", encoding="utf-8") as f:
    lines = f.readlines()
    # 跳过表头，从第2行开始（行号从1开始计数）
    for i, line in enumerate(lines[1:], start=2):
        try:
            record = parse_student(line)
            valid_records.append(record)
        except ValueError as e:
            invalid_records.append({"line": i, "content": line.strip(), "reason": str(e)})

print("\n有效记录数:", len(valid_records))
print("无效记录数:", len(invalid_records))

print("\n无效记录详情:")
for item in invalid_records:
    print(f"  行号 {item['line']}: {item['content']} -> 原因: {item['reason']}")

print("\n有效记录示例 (前3条):")
for rec in valid_records[:3]:
    print(rec)
```

```python
#任务5
report_path = os.path.join("exp01_data", "report.txt")

# 计算平均分和及格率
total_score = sum(rec["score"] for rec in valid_records)
avg_score = total_score / len(valid_records) if valid_records else 0
pass_count = sum(1 for rec in valid_records if rec["score"] >= 60)
pass_rate = pass_count / len(valid_records) if valid_records else 0

report_content = (
    f"有效记录数: {len(valid_records)}\n"
    f"无效记录数: {len(invalid_records)}\n"
    f"平均分: {avg_score:.2f}\n"
    f"及格率: {pass_rate:.2%}\n"
)

# 使用 w 模式（覆盖）
with open(report_path, "w", encoding="utf-8") as f:
    f.write(report_content)
print("\n--- 使用 w 模式写入 report.txt ---")
print("w 模式会覆盖已有内容。")

# 再次使用 a 模式（追加）
with open(report_path, "a", encoding="utf-8") as f:
    f.write("\n--- 追加内容（a 模式）---\n")
    f.write("a 模式不会覆盖，而是在末尾追加。\n")

# 读取并显示
with open(report_path, "r", encoding="utf-8") as f:
    print("\nreport.txt 最终内容:")
    print(f.read())
```

## 实验五：类、继承与对象复制

```python
#任务1
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"姓名: {self.name}, 成绩: {self.score}"

s = Student("Alice", 85)
print(s)
print("是否及格:", s.is_passed())
```

```python
#任务2
students = [Student(rec["name"], rec["score"]) for rec in valid_records]

print(f"共 {len(students)} 条有效记录：")
for stu in students:
    print(stu, "| 及格状态:", "及格" if stu.is_passed() else "不及格")
```

```python
#任务3
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)      # 初始化父类属性
        self.research_topic = research_topic

    def __str__(self):
        return f"研究生 姓名: {self.name}, 成绩: {self.score}, 研究方向: {self.research_topic}"

g = GraduateStudent("David", 88, "数据分析")
print(g)
print("是否及格:", g.is_passed())
print("是否属于Student:", isinstance(g, Student))
```

```python
#任务4
import copy

def make_obj():
    return {"name": "Alice", "scores": [85, 90]}

# 1. 直接赋值
obj1 = make_obj()
alias = obj1
alias["scores"].append(100)
print("直接赋值 -> 原对象:", obj1)

# 2. 浅拷贝
obj2 = make_obj()
shallow = copy.copy(obj2)
shallow["scores"].append(100)
print("浅拷贝   -> 原对象:", obj2)

# 3. 深拷贝
obj3 = make_obj()
deep = copy.deepcopy(obj3)
deep["scores"].append(100)
print("深拷贝   -> 原对象:", obj3)
```
